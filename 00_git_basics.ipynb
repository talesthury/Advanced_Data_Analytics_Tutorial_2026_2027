Git sounds complicated, but in daily life you only need **7 commands**. This notebook shows them in the four situations where you will actually use them.

| Situation | Commands |
|---|---|
| **A.** Get a repo from GitHub onto your laptop | `git clone` |
| **B.** Get the newest files | `git pull` |
| **C.** You worked on a file and want to save it on GitHub | `git add` · `git commit` · `git push` |
| **D.** You have a folder on your laptop and want to put it on GitHub | `git init` · `git remote add` · and the three of C |

Plus `git status`, the command you will use most: *"what is going on right now?"*

## The picture to keep in mind

```
      GitHub (online)
   +------------------+
   |   your repo      |
   +------------------+
        |        ^
  clone |        | push
  pull  |        |
        v        |
   +------------------+
   |  your laptop     |   <- the folder you open in VS Code
   +------------------+
```

- **clone**: download the repo the first time
- **pull**: download what is new
- **push**: upload what you changed

Everything else is preparation for these three.

## A · Get a repo onto your laptop (fork + clone)

**When:** a new course starts and you want the course files in VS Code.

**1. Fork on GitHub.** Open the course repo and click **Fork → Create fork**. Now you have your own copy online, under **your** username.

**2. Make one dedicated folder for the course** on your laptop, for example:
```
C:\Users\you\Documents\DataMining
```

**3. Clone your fork into that folder.** On your fork, click the green **Code** button, copy the link, and in the terminal:
```bash
cd C:\Users\you\Documents\DataMining
git clone https://github.com/YOUR-USERNAME/Advanced_Data_Analytics_Tutorial_2026_2027.git
```

**4. Open it in VS Code:** *File → Open Folder* → choose the new folder.

> **Be careful when cloning**
> - Copy the link of **your fork** (with your username), not the original repo.
> - Do not clone into OneDrive, iCloud or Dropbox folders.
> - Do not clone inside another repo. Keep course repos **next to each other**:
>   ```
>   DataMining/
>   ├── lecture-repo/
>   └── Advanced_Data_Analytics_Tutorial_2026_2027/
>   ```
> - You clone **only once**. After that, you use `pull`.

## B · Get the newest files (every week)

**When:** a new notebook was published.

**1. On GitHub**, open your fork and click **Sync fork → Update branch**.

**2. On your laptop**, in the terminal of the course folder:
```bash
git pull
```

That's it. The new files are now in your folder.

> **Tip: work on a copy.** Before you fill in a notebook, copy it into a folder `my_work/` and work there.
> If you edit the original and the tutor later fixes that same file, `git pull` stops with an error, because Git does not know which version to keep.

## C · You worked on a file and want to save it on GitHub

**When:** you finished an exercise and want it online (backup, or to share).

Think of it as sending a parcel:

| Step | Command | Parcel |
|---|---|---|
| 1 | `git status` | look at what changed |
| 2 | `git add my_work/02_outliers_MINE.ipynb` | put the file into the box |
| 3 | `git commit -m "Finished outlier exercise"` | close the box and write a label |
| 4 | `git push` | send the box to GitHub |

```bash
git status
git add my_work/02_outliers_MINE.ipynb
git commit -m "Finished outlier exercise"
git push
```

`git add .` (with a dot) puts **all** changed files into the box at once.

## D · You have a folder on your laptop and want to put it on GitHub

**When:** you started a project on your own (for example a script for your thesis) and now want it on GitHub.

**1. On GitHub:** click **+ → New repository**, give it a name, and **do not** add a README. Click **Create repository**. Copy the link it shows.

**2. In the terminal, inside your project folder:**
```bash
cd C:\Users\you\Documents\my_project
git init
git add .
git commit -m "First commit"
git branch -M main
git remote add origin https://github.com/YOUR-USERNAME/my_project.git
git push -u origin main
```

| Command | What it does |
|---|---|
| `git init` | turns the folder into a Git folder (once) |
| `git add .` / `git commit` | the parcel from C |
| `git branch -M main` | names the main line `main` |
| `git remote add origin <link>` | tells the folder where its GitHub repo is (once) |
| `git push -u origin main` | the first upload; after that, just `git push` |

GitHub shows exactly these commands on the page of a new, empty repository, so you can copy them from there.

## The 7 commands on one page

| Command | In one sentence |
|---|---|
| `git clone <link>` | download a repo the first time |
| `git status` | what changed? |
| `git add <file>` | choose what to save |
| `git commit -m "message"` | save it with a label |
| `git push` | upload to GitHub |
| `git pull` | download what is new |
| `git init` | make a normal folder a Git folder |

## When something goes wrong

| You see | It means | Do this |
|---|---|---|
| `not a git repository` | the terminal is in the wrong folder | `cd` into the course folder |
| `rejected ... fetch first` | GitHub has something you don't have yet | `git pull`, then `git push` |
| `local changes would be overwritten` | you edited an original file | copy it into `my_work/`, then `git restore <file>`, then `git pull` |
| `nothing to commit` | there is nothing new to save | nothing, all good |
